```python
import pandas as pd
import numpy as np
```

```python
df_ASV = pd.read_csv('Data/ASV_cleaned.csv')

df_net = pd.read_csv('Data/df_net.csv')

all_id = np.unique(df_ASV['Person ID'])
ids = [hex(x) for x in np.random.choice(10**6, len(all_id), replace = False)]

Mapper = dict(zip(all_id, ids))

df_net.pid = df_net.pid.map(lambda x: Mapper[x])
df_net.pid2 = df_net.pid2.map(lambda x: Mapper[x])

df_ASV['Person ID'] = df_ASV['Person ID'].map(lambda x: Mapper[x])
```

```python
df_net.to_csv('Data/df_net.csv', index = False)
```

```python
df_ASV.to_csv('Data/ASV_cleaned.csv', index = False)
```
